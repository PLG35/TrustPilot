## 6. Distribution des notes

```python
print("Kaggle")
display(df_kaggle["rating"].value_counts(normalize=True).sort_index() * 100)

print("\nHugging Face")
display(df_hf["stars"].value_counts(normalize=True).sort_index() * 100)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.countplot(x="rating", data=df_kaggle, palette="RdYlGn", hue="rating", legend=False, ax=axes[0])
axes[0].set_title("Distribution des notes - Kaggle")
axes[0].set_xlabel("Note")
axes[0].set_ylabel("Nombre d'avis")

sns.countplot(x="stars", data=df_hf, palette="RdYlGn", hue="stars", legend=False, ax=axes[1])
axes[1].set_title("Distribution des notes - Hugging Face")
axes[1].set_xlabel("Note")
axes[1].set_ylabel("Nombre d'avis")

plt.tight_layout()
plt.show()
```

### Interprétation

La distribution des notes montre une différence importante entre les deux datasets. **Kaggle** est très fortement concentré sur les avis à **5 étoiles**, tandis que **Hugging Face** présente une répartition beaucoup plus équilibrée entre les cinq niveaux de notes.

Cette uniformité ne doit cependant pas être considérée comme plus représentative de Trustpilot. À titre de repère externe, Trustpilot indiquait dans son **Transparency Report 2022** que **72,51 %** des avis soumis à la plateforme étaient des avis à **5 étoiles**, contre **13,56 %** à **1 étoile**. La distribution Hugging Face apparaît donc atypique et son origine doit être étudiée dans les sections suivantes.

Source : Trustpilot, *6 things that might surprise you about Trustpilot* (données du Transparency Report 2022).

```python
from scipy.stats import chi2_contingency

table_notes = pd.DataFrame({
    "Kaggle": df_kaggle["rating"].value_counts().sort_index(),
    "Hugging Face": df_hf["stars"].value_counts().sort_index()}).fillna(0)

resultats_chi2 = chi2_contingency(table_notes)

statistique = resultats_chi2[0]
p_valeur = resultats_chi2[1]

print("Statistique du test :", statistique)
print("P-valeur :", p_valeur)

if p_valeur < 0.05:
    print("H0 rejetée : la distribution des notes n'est pas indépendante du dataset.")
    print("Conclusion : les distributions de notes sont statistiquement différentes.")
else:
    print("H0 non rejetée : on ne met pas en évidence de dépendance entre la distribution des notes et le dataset.")
    print("Conclusion : on ne met pas en évidence de différence statistiquement significative.")
```

### Interprétation du test du Chi²

Le test du Chi² permet de vérifier statistiquement l'écart observé sur les graphiques. La p-value étant très inférieure à **0,05**, les distributions de notes des deux corpus sont considérées comme statistiquement différentes. Ce test confirme la différence, mais n'en explique pas la cause : le mécanisme d'échantillonnage sera étudié ensuite.
Le graphique montre que les distributions sont différentes.
Le test du Chi² confirme que cette différence est statistiquement significative.
cette différence ne pourrait pas etre due au hasard d’échantillonnage
- *H0* : la distribution des notes est indépendante du dataset.
- *H1* : la distribution des notes dépend du dataset.

## 9. Vérification du plafonnement

```python
repartition_kaggle = (df_kaggle.groupby(["name", "rating"]).size().unstack(fill_value=0))

display(repartition_kaggle.max())
```

```python
repartition_hf = (df_hf.groupby(["company", "stars"]).size().unstack(fill_value=0))

display(repartition_hf.max())
```

```python
entreprises_100 = avis_par_entreprise_hf[avis_par_entreprise_hf == 100].index
repartition_100 = repartition_hf.loc[entreprises_100]

print("Entreprises avec 100 avis et 20 avis pour chacune des 5 notes :",
      (repartition_100 == 20).all(axis=1).sum())
```

### Interprétation

L’analyse confirme effectivement l’existence d’un **plafonnement dans les deux datasets**. Dans **Kaggle**, le nombre d’avis est limité à **20 avis au total par entreprise**. Dans **Hugging Face**, le plafonnement est différent : il est limité à **20 avis par note et par entreprise**, soit jusqu’à 100 avis par entreprise. Ce mécanisme explique en partie la répartition plus équilibrée des étoiles dans Hugging Face, qui ne reflète donc pas une distribution naturelle des notes.
La distribution des étoiles Hugging Face est donc artificiellement équilibrée.

## 11. Analyse temporelle de Kaggle

```python
df_kaggle["reviewed_at"] = pd.to_datetime(df_kaggle["reviewed_at"])
df_kaggle["scraped_at"] = pd.to_datetime(df_kaggle["scraped_at"])

avis_par_annee = df_kaggle["reviewed_at"].dt.year.value_counts().sort_index()

plt.figure(figsize=(10, 5))

sns.barplot(x=avis_par_annee.index, y=avis_par_annee.values)
sns.lineplot(x=range(len(avis_par_annee)), y=avis_par_annee.values, color = "red", marker="o")

plt.title("Évolution du nombre d'avis Kaggle par année")
plt.xlabel("Année")
plt.ylabel("Nombre d'avis")
plt.show()
```

```python
stats_temporelles_kaggle = (df_kaggle.groupby("name")["reviewed_at"].agg(["min", "max", "count"]))

stats_temporelles_kaggle["duree_jours"] = (stats_temporelles_kaggle["max"]
                                           - stats_temporelles_kaggle["min"]).dt.days

display(stats_temporelles_kaggle["duree_jours"].describe())
```

```python
df_kaggle["reviewed_at"] = pd.to_datetime(df_kaggle["reviewed_at"], utc=True)
df_kaggle["scraped_at"] = pd.to_datetime(df_kaggle["scraped_at"], utc=True)
print("Kaggle : ")
entreprises_20 = avis_par_entreprise_kaggle[avis_par_entreprise_kaggle == 20].index

df_entreprises_20 = df_kaggle[df_kaggle["name"].isin(entreprises_20)]

dernier_avis = df_entreprises_20.groupby("name")["reviewed_at"].max()
date_scraping = df_kaggle["scraped_at"].max()

ecart_scraping = (date_scraping - dernier_avis).dt.days

print("Nombre d'entreprises ayant exactement 20 avis :")
print(len(entreprises_20))

print("\nDate de scraping du dataset :")
print(date_scraping)

print("\nStatistiques sur le nombre de jours entre le dernier avis de chaque entreprise")
print("et la date de scraping , pour les entreprises ayant exactement 20 avis : ")

print("\n")
display(ecart_scraping.describe())
print("La valeur minimale de -1 jour s’explique probablement par la précision horaire des dates",
": la date de scraping est enregistrée à minuit, tandis que certains avis ont été publiés plus tard",
"au cours de la même journée. Le calcul en jours entiers transforme alors cette différence de quelques",
"heures en -1.")


print("\nPour les entreprises ayant exactement 20 avis :")
print("\nEn moyenne, le dernier avis a été publié", round(ecart_scraping.mean(), 2), "jour(s) avant le scraping.")

print("La médiane est de", ecart_scraping.median(),"jour(s).")

print("75 % de ces entreprises ont leur dernier avis à", ecart_scraping.quantile(0.75), "jour(s) ou moins de la date de scraping.")

print("L'écart maximum observé est de", ecart_scraping.max(), "jour(s).")
```

### Interprétation

L’analyse temporelle montre que les avis Kaggle couvrent plusieurs années, mais que pour les entreprises ayant exactement **20 avis**, le dernier avis collecté est généralement très proche de la date de scraping. L’écart médian est d’environ **1 jour**, et **75 %** de ces entreprises ont leur dernier avis à **3 jours ou moins** de la date de collecte. Cela suggère fortement un **biais de récence** dans le dataset Kaggle, sans toutefois permettre d’affirmer avec certitude qu’il s’agit exactement des 20 derniers avis publiés.

### Point de vigilance sur la qualité des données

Le **volume de valeurs manquantes** doit être considéré comme un critère de qualité parmi d'autres. Un taux élevé de NA sur des variables importantes augmente le besoin de prétraitement et peut entraîner une perte ou une transformation de l'information. À l'inverse, **0 % de valeurs manquantes n'est pas automatiquement une preuve de meilleure qualité**, surtout lorsque les étapes de nettoyage du dataset ne sont pas documentées.

### Conclusion

L'analyse comparative montre que les deux corpus sont largement distincts : seuls **7 commentaires communs** ont été identifiés. Le corpus Hugging Face ne constitue donc pas une simple extension de Kaggle.

La distribution des notes de **Kaggle** est fortement dominée par les **5 étoiles**, ce qui semble **plus proche de la forme générale de la distribution publiée par Trustpilot** que la répartition beaucoup plus uniforme de Hugging Face. Cela ne permet toutefois pas d'affirmer que Kaggle est représentatif de Trustpilot : le corpus reste limité à **3 698 avis**, **214 entreprises** et un maximum observé de **20 avis par entreprise**, avec en plus un biais de récence probable.

**Hugging Face** est beaucoup plus riche pour la modélisation textuelle : **123 181 avis**, **1 680 entreprises**, **22 catégories**, des textes globalement plus longs et davantage d'exemples pour chaque niveau de note, notamment les notes faibles. Son principal défaut est son mécanisme d'échantillonnage : le nombre d'avis observé est plafonné à **20 avis par note et par entreprise**, ce qui rend la distribution globale des étoiles artificielle et concentre de nombreuses moyennes d'entreprises autour de **3/5**.

Le test du **Chi²** confirme que les distributions de notes des deux corpus sont statistiquement différentes, mais il ne suffit pas à expliquer cette différence ni à juger leur représentativité.

Concernant la qualité des données, Kaggle contient quelques valeurs manquantes, essentiellement dans `description`, sans impact direct majeur sur les variables utilisées pour le NLP. Hugging Face ne contient aucun NA, mais cette absence parfaite peut résulter d'un nettoyage réalisé en amont ; les éventuelles suppressions ou imputations n'étant pas documentées, elles doivent être signalées comme une limite méthodologique.

Ainsi :
- **Hugging Face** est retenu comme corpus principal pour le NLP, grâce à son volume et à sa diversité textuelle ;
- **Kaggle** est conservé comme corpus secondaire pour la temporalité, le contrôle externe, la comparaison de robustesse et comme point de comparaison dont la distribution des notes semble plus proche du profil global publié par Trustpilot.

Les biais de sélection et les étapes de préparation non documentées devront être pris en compte dans l'interprétation des résultats.
